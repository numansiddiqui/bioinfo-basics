```python
from Bio import SeqIO

path = "data/sequence.fasta"
records = list(SeqIO.parse(path, "fasta"))

print("Number of sequences:", len(records))
first = records[0]
print("ID:", first.id)
print("Description:", first.description)
print("Length:", len(first.seq))
print(records[0].seq)
```

```python
def count_atgc(seq):
    seq_str = str(seq).upper()
    counts = {}
    for base in "ATGC":
        counts[base] = seq_str.count(base)    
    return counts
```

```python
print(count_atgc(first.seq))
```

```python
def gc_count(seq):
    seq_str = str(seq).upper()
    counts = {}
    for gc in "GC":
        counts[gc] = seq_str.count(gc)

        
    counts_sum = sum(counts.values())
    percent_gc = (counts_sum/len(first.seq))
    return percent_gc
```

```python
print(gc_count(first.seq))
```
